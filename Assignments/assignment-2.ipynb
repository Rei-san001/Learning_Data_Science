```python
# 1. Write a Python program that takes a list of numbers as input, removes duplicates using a suitable list method, and returns a dictionary containing the original list, the unique values, and their count. Use a function with parameters and a return statement to perform this task.


def no_duplicate(numbers):
    num_unique = []
    for i in numbers:
        if i not in num_unique:
            num_unique.append(i)
    return {
        "original": numbers,
        "unique": num_unique,
        "count": len(num_unique)
    }

numbers = []
my_input = input()

for i in my_input.split():
    numbers.append(int(i))

result = no_duplicate(numbers)

print("Original list:", result["original"])
print("Unique values:", result["unique"])
print("Count:", result["count"])
```

```python
# 2. Create a function that accepts two sets as parameters and returns their union, intersection, and difference. Use keyword arguments with default parameter values so the function can work even if one of the sets is not provided by the user. Display the results clearly.


def diff_sets(setA=set(), setB=set()):
    union = setA | setB
    intersection = setA & setB
    difference = setA - setB

    return union, intersection, difference

#setA and setB
union, intersection, difference = diff_sets(setA={1,2,3,4}, setB={5,6,7,8})

print("UNION: ", union)
print("INTERSECTION: ", intersection)
print("DIFFERENCE: ", difference)

print(" ")

#setA only - to check the deafult set property
union, intersection, difference = diff_sets(setA={1,2,3,4})
print("UNION_A: ", union)
print("INTERSECTION_A: ", intersection)
print("DIFFERENCE_A: ", difference)

print(" ")

#setB only - to check the deafult set property
union, intersection, difference = diff_sets(setB={5,6,7,8})
print("UNION_B: ", union)
print("INTERSECTION_B: ", intersection)
print("DIFFERENCE_B: ", difference)
```

```python
# 3. Define a tuple containing mixed data types, unpack its values into separate variables, and compare them with another tuple using comparison operators. Then, explain in code comments the main difference between lists and tuples in Python.


tuple1 = ("Kiwi", 10, 3.9, True)
fruit_name, quantity, pH_level, is_fruit = tuple1

tuple2 = ("Tomato", 10, 4.3, False)
fruit_name1, quantity1, pH_level2, is_fruit1 = tuple2

# comparing tuples

print(tuple1 == tuple2)
print(tuple1 != tuple2)
print(tuple1 > tuple2)
print(tuple1 < tuple2)
print(tuple1 >= tuple2)
print(tuple1 <= tuple2)

print(" ")

#FOR comparing values of tuples

tuple_pair = [
    ("fruit_name", fruit_name, fruit_name1),
    ("quantity", quantity, quantity1),
    ("pH_level", pH_level, pH_level2),
    ("is_fruit", is_fruit, is_fruit1),
]
for label, a, b in tuple_pair:
    print(f"{label}: equal:{a == b}  not_equal:{a != b}  less_than:{a < b}  greater_than:{a > b}  less_equal:{a <= b}  greater_equal:{a >= b}")


print(" ")

"""
LIST:
1. List is mutable, means we can change the content of the list after it is created
2. Syntax: [1, 2, 3, 4, 5]
3. Bit slower
4. Many built-in methods available for list

TUPLE:
1. Tuple is immutable, means we cannot change the content of the tuple after it is created
2. Syntax: (1, 2, 3, 4, 5)
3. Bit faster
4. Lesser built-in methods available for tuple


"""
```
