```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv')
df_laptops.head(5)
```

## 1. Filtering DF using multiple conditions 

```python
# apple laptops that costs more than 2000

(df_laptops['Company'] == 'Apple') & (df_laptops['Price_euros'] > 2000)
```

## 2.1 & (and)

```python
# filtering the DF

df_laptops[(df_laptops['Company'] == 'Apple') & (df_laptops['Price_euros'] > 2000)]
```

## 2.2 | (or)

```python
(df_laptops['Company'] == 'Apple') | (df_laptops['Price_euros'] > 2000)
```

```python
df_laptops[(df_laptops['Company'] == 'Apple') | (df_laptops['Price_euros'] > 2000)]
```
