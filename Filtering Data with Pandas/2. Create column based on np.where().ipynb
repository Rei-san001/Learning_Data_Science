# 1. Create column using np.where()

```python
import numpy as np
```

```python
# creating array based on price tiers

np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Affordable')
```

```python
# set it to a new column

df_laptops['Price_tier'] = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Affordable')
```
