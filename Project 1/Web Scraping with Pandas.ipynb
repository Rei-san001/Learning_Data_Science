# 2. Read a .csv file from multiple URLs with Pandas

- https://football-data.co.uk/mmz4281/2526/E0.csv
- https://football-data.co.uk/mmz4281/2526/E1.csv

**Link: root + season + league**

```python
#link structure 

"https://football-data.co.uk/mmz4281/" + "2526" + "/" + "E0" + ".csv"
```

```python
# create a root variable 
root = "https://football-data.co.uk/mmz4281/"
```

## 2.1 Multiple Leagus

- https://football-data.co.uk/mmz4281/2526/E0.csv
- https://football-data.co.uk/mmz4281/2526/E1.csv
- https://football-data.co.uk/mmz4281/2526/E2.csv
- https://football-data.co.uk/mmz4281/2526/E3.csv
- https://football-data.co.uk/mmz4281/2526/EC.csv

```python
# creating a list of leagus
leagues = ['E0', 'E1', 'E2', 'E3', 'EC']

frames = [] # create an empty list 
 

# looping through leagues, read multiple csv's and append to a list 
for league in leagues: 
    df = pd.read_csv(root + '2526' + '/' + league + '.csv')
    frames.append(df)   #to store each dataframe in the list
```

```python
# length of frames list
len(frames)
```

```python
frames[4]
```

## 2.2 Multiple Seasons

```python
for season in range(20, 26):
    print(str(season) + str(season+1)) #to show whats actually happening in the loop
```

```python
# creating a list of leagus
leagues = ['E0', 'E2', 'E3']

frames = [] # create an empty list 


# loop through range of seasons and insert new column
for league in leagues: 
    for season in range(20, 26):
        try:
            df = pd.read_csv(root + str(season) + str(season+1) + '/' + league + '.csv')
        except Exception:
            print('Failed:', league, season)
            continue
        df.insert(0, 'Season', season) #to insert a new column at the first position
        frames.append(df)
```

```python
# 6 season times 3 leagues
len(frames)
```
