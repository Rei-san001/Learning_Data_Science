# 1. Creating a dataframe using array

### 1.1 Option-1

```python
# creating an array
data = np.array([[1,4], [2,5], [3,6]])
```

```python
# creating a dataframe 
df = pd.DataFrame(data, 
             index=['row1', 'row2', 'row3'], 
             columns=['col1', 'col2']
            )
```

```python
# showing the dataframe
df
```

### 1.2 Option-2

# 2. Creating a DataFrame from dictinary 

```python
# lists used for this one
states = ['California', 'Texas', 'New York', 'Florida', 'Illinois']
population = [22.2, 29.0, 19.8, 21.5, 12.7]
```

```python
# storing lists in the dictionary
dict_states = {'States':states,
               'Population':population
               }
```

```python
# creating the dataframe from the dictionary
df_population = pd.DataFrame(dict_states)
```

```python
df_population
```
