```python
import pandas as pd
```

```python
df_exams = pd.read_csv('StudentsPerformance.csv')
df_exams
```

# 1. Create Index

```python
import numpy as np
import random
```

```python
# creating non repititive values for the index cuz index in a dataframe doesnt allow repititive values
new_index = np.arange(0, 1000)
```

```python
# shuffling the new_index
random.shuffle(new_index)
```

```python
# creating a new column with those random values
df_exams['new_index_column'] = new_index
df_exams
```

# 2. Set Index

    > set_index() helps us to use a column as index for the dataframe

```python
# set new_index column as the index
df_exams.set_index('new_index_column', inplace  = True)
df_exams
```

# 3. Sort Index

    > It sorts the new_index_column that we created for the index

```python
# sort the dataframe by new index
df_exams.sort_index()
```

```python
df_exams.sort_index(ascending = False)
```
