```python
import pandas as pd
```

```python
# reading and showing the dataframe 
df_exams = pd.read_csv('StudentsPerformance.csv')
df_exams
```

# 1. Sort a DataFrame

```python
# sort by one column 
df_exams.sort_values(by= 'math score') #we can use it without the "by="
```

```python
# sort descending by one column 
df_exams.sort_values(by="math score", ascending=False)
```

```python
# sort descending by multiple columns 
df_exams.sort_values(by=["math score", "reading score"], ascending=False)
# but this didn't update the dataframe, it's just a copy
```

```python
# sort descending by multiple columns update dataframe
df_exams.sort_values(by=["math score", "reading score"], ascending=False,
                     inplace = True) #to update the dataframe we need to use inplace = True
```

```python
df_exams
```

```python
# sort descending with a key function
df_exams.sort_values('race/ethnicity', ascending=True,
                     key=lambda col:col.str.lower())    #lambda function is used to convert the column values to lowercase before sorting, ensuring a case-insensitive sort.
```
