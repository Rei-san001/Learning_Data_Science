### 1.2 Adding a new column with an array

the dataframe has 1000 rows, which is why we need to create an array with 1000 elements 

```python
# import numpy
import numpy as np
```

```python
# array of 1000 elements 
language_score = np.arange(0, 1000)
```

```python
len(language_score)
```

```python
# adding new column using the array
df_exams['language score'] = language_score
df_exams.head()
```

```python
# create random integer numbers between 1 to 100
int_language_score = np.random.randint(1, 100, size = 1000)
```

```python
len(int_language_score)
```

```python
# min value = inclusive and high value = exclusive
min(int_language_score)
max(int_language_score)
```

```python
# adding the new column using array
df_exams['language score'] = int_language_score
df_exams
```
